```python
label_counts = {}

for file in csv_files:
    print(f"Reading: {file.name}")

    labels = pd.read_csv(
        file,
        usecols=[" Label"]
    )[" Label"].str.strip()

    counts = labels.value_counts()
    label_counts[file.name] = counts

    print(counts)
    print("-" * 60)
```

```python
total_label_counts = pd.concat(
    label_counts.values()
).groupby(level=0).sum()

print("\n===== TOTAL CICIDS2017 LABEL DISTRIBUTION =====\n")
print(total_label_counts.sort_values(ascending=False))
```

```python
missing_summary = {}

for file in csv_files:
    print(f"Checking: {file.name}")

    df_temp = pd.read_csv(file)

    missing = df_temp.isnull().sum().sum()

    missing_summary[file.name] = missing

    print("Missing cells:", missing)
    print("-" * 60)

print("\n===== TOTAL =====")
print(sum(missing_summary.values()))
```

```python
infinite_summary = {}

for file in csv_files:
    print(f"Checking: {file.name}")

    infinite_total = 0

    for chunk in pd.read_csv(file, chunksize=100_000):
        numeric_chunk = chunk.select_dtypes(include=np.number)

        infinite_total += np.isinf(
            numeric_chunk.to_numpy()
        ).sum()

    infinite_summary[file.name] = infinite_total

    print("Infinite values:", infinite_total)
    print("-" * 60)

print("\n===== TOTAL INFINITE VALUES =====")
print(sum(infinite_summary.values()))
```

```python
duplicate_counts = {}

for file in csv_files:
    print(f"Checking: {file.name}")

    seen_rows = set()
    duplicate_count = 0

    for chunk in pd.read_csv(file, chunksize=100_000):
        # Normalize column whitespace for comparison only
        chunk.columns = chunk.columns.str.strip()

        # Convert each row to a hashable representation
        row_hashes = pd.util.hash_pandas_object(
            chunk,
            index=False
        )

        for h in row_hashes:
            h = int(h)

            if h in seen_rows:
                duplicate_count += 1
            else:
                seen_rows.add(h)

    duplicate_counts[file.name] = duplicate_count

    print("Duplicate rows:", duplicate_count)
    print("-" * 60)

print("\n===== TOTAL DUPLICATE ROWS =====")
print(sum(duplicate_counts.values()))
```

```python
dtype_summary = {}

for file in csv_files:
    print(f"Checking: {file.name}")

    df_temp = pd.read_csv(file, nrows=1000)

    # Remove whitespace from column names
    df_temp.columns = df_temp.columns.str.strip()

    counts = df_temp.dtypes.value_counts()

    dtype_summary[file.name] = counts

    print(counts)
    print("-" * 60)
```

```python
from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path.cwd().parent

FINAL_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "CICIDS2017_final"
)

final_files = sorted(FINAL_DIR.glob("*.csv"))

print("Final files:", len(final_files))

for file in final_files:
    print(file.name)
```

```python
for file in final_files:
    df_check = pd.read_csv(file, nrows=5)
    df_check.columns = df_check.columns.str.strip()

    print(
        file.name,
        "->",
        len(df_check.columns),
        "columns"
    )
```

```python
for file in final_files:
    print(f"\nChecking: {file.name}")

    missing = 0
    infinite = 0

    for chunk in pd.read_csv(file, chunksize=100_000):

        missing += chunk.isnull().sum().sum()

        numeric = chunk.select_dtypes(include=np.number)

        infinite += np.isinf(
            numeric.to_numpy()
        ).sum()

    print("Missing:", missing)
    print("Infinite:", infinite)
```

```python
final_label_counts = {}

for file in final_files:

    labels = pd.read_csv(
        file,
        usecols=["Label"]
    )["Label"].str.strip()

    final_label_counts[file.name] = labels.value_counts()


total_final_labels = pd.concat(
    final_label_counts.values()
).groupby(level=0).sum()

print(
    total_final_labels
    .sort_values(ascending=False)
)
```
